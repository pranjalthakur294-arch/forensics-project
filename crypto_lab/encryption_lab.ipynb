# Forensic Evidence Encryption & Integrity Lab
This notebook demonstrates forensic-grade encryption, authenticated integrity checking (AES-256-GCM), key derivation (PBKDF2-HMAC-SHA256), and chain-of-custody tracking using `crypto_tool.py`.

```python
from crypto_tool import encrypt_file, decrypt_file, compute_sha256, verify_integrity, IntegrityError
import os

# 1. Create a sample piece of digital evidence
sample_file = 'evidence_note.txt'
with open(sample_file, 'w') as f:
    f.write('Case #2026-0042: Suspect communications and timestamp data.')

print('Original SHA-256:', compute_sha256(sample_file))
```

```python
# 2. Encrypt using AES-256-GCM with PBKDF2 key derivation
password = 'CaseOfficerSecretPassword99!'
meta = encrypt_file(sample_file, password=password)
print('Encrypted output metadata:')
for k, v in meta.items():
    print(f'  {k}: {v}')
```

```python
# 3. Test tamper-resistance: Attempt decryption with wrong password
try:
    decrypt_file('evidence_note.txt.enc', password='WrongPassword!')
except IntegrityError as e:
    print('Security check passed: Tampering/incorrect password blocked ->', e)
```

```python
# 4. Decrypt successfully with correct credentials
recovered_file = decrypt_file('evidence_note.txt.enc', output_path='evidence_note_recovered.txt', password=password)
with open(recovered_file, 'r') as f:
    print('Recovered Content:', f.read())

is_valid = verify_integrity(recovered_file, 'evidence_note.txt.enc.evidence.txt')
print('Chain of custody intact:', is_valid)
```
